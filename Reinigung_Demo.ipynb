### Teil 2: Leere Zellen

Wir schauen zuerst, was uns *info()* über das DataFrame sagt.

```python
workout.info()
```

Es gibt drei verschiedene Varianten, wie wir die leeren Zellen behandeln können.
1. Wir entfernen alle Zeilen mit einer leeren Zelle.
2. Wir schreiben einen bestimmten Wert in die leeren Zellen.
3. Wir schreiben den Median, Durchschnitt oder Mode (der häufigste Wert) dieser Spalte in die Zelle hinein.

Diese drei Varianten probieren wir nun aus.

Bevor wir diese drei Varianten jedoch ausprobieren, machen wir eine Kopie von unserem DataFrame und speichern Sie in der Variable *workout_vor_leere_zellen*. Diese Kopie ist nur für die Demo nötig, damit wir auf das gleiche DataFrame verschieden Arten testen können, um leere Zellen aufzufüllen. In der Praxis und auch beim Lösen der Aufgaben später, müssen wir das nicht machen.

```python
workout_vor_leere_zellen = workout.copy() # diese Kopie brauchen wir nur für die Demo
```

#### 1. Leere Zeilen ganz entfernen

```python
workout.dropna(inplace = True) # entfernt die Zeilen mit leeren Zellen
workout
```

```python
workout.info()
```

Datensätze sind meistens so gross, dass es keine Rolle für das Resultat spielt, ob eine Zeile mehr oder weniger vorhanden ist. Das Entfernen einer ganzen Zeile ist somit nicht schön, aber in den meisten Fällen in Ordnung.

#### 2. Spezifischer Wert in Zelle füllen

Eine andere Möglichkeit ist es, die leeren Zellen mit einem bestimmten Wert aufzufüllen. Das ist natürlich nur sinnvoll, wenn man die Daten gut kennt und weiss, welcher Wert passen könnte.

```python
workout = workout_vor_leere_zellen.copy() # nur wegen der Demo benötigt, setzt das DataFrame wieder zurück auf den Zustand vor dem auffüllen der leeren Zellen
workout.fillna(230, inplace = True)
workout
```

In der Spalte Datum steht beim Eintrag 22. nun auch der Wert 230. Das ergibt wenig Sinn. Wir versuchen es mit dem gezielten Auffüllen noch einmal und setzten den Wert 230 nur in der Spalte *kalorien* ein. 

```python
workout = workout_vor_leere_zellen.copy() # nur wegen der Demo benötigt, setzt das DataFrame wieder zurück auf den Zustand vor dem auffüllen der leeren Zellen
workout["kalorien"].fillna(230, inplace = True)
workout
```

#### 3. Zelle mit Median, Durchschnitt oder Mode der jeweiligen Spalte füllen

```python
workout = workout_vor_leere_zellen.copy() # nur wegen der Demo benötigt, setzt das DataFrame wieder zurück auf den Zustand vor dem auffüllen der leeren Zellen
x_mean = workout["kalorien"].mean()
print(x_mean)

workout["kalorien"].fillna(x_mean, inplace = True)
workout
```

Anstatt den Durchschnitt hätten wir hier auch den Median oder den am häufigsten vorkommenden Wert zum Ersetzen verwenden können.

```python
workout = workout_vor_leere_zellen.copy() # nur wegen der Demo benötigt, setzt das DataFrame wieder zurück auf den Zustand vor dem auffüllen der leeren Zellen
x_median = workout["kalorien"].median()
print(x_median)
workout["kalorien"].fillna(x_median, inplace = True)
workout
```

```python
workout = workout_vor_leere_zellen.copy() # nur wegen der Demo benötigt, setzt das DataFrame wieder zurück auf den Zustand vor dem auffüllen der leeren Zellen
x_mode = workout["kalorien"].mode()[0] # gibt den in der Spalte kalorien am häufigsten vorkommenden Wert zurück
print(x_mode)
workout["kalorien"].fillna(x_mode, inplace = True)
workout
```

### Teil 4: Falsche Daten

Daten werden häufig erfasst, in dem Menschen Formulare ausfüllen. Dabei kommt es schnell zu Tippfehlern oder die Einheit, in welcher die Daten eingegeben werden sollen, wird überlesen. Schauen wir uns doch noch einmal unseren Datensatz an.

```python
workout
```

Die Dauer des Workouts in Zeile 7 sticht ziemlich heraus. Alle anderen Workouts waren zwischen 45 und 60 Minuten. Der zugehörige Wert in der Spalte *Kalorien* sieht auch normal aus und ist nicht viel höher als die anderen Werte. Vermutlich wurde nur eine 0 zu viel eingetippt.

```python
workout.loc[7, 'dauer'] = 45 # ändert den Wert in der Zeile mit dem Index 7 und der Spalte 'dauer' zu 45
workout
```

Das ging sehr flott. Wenn unser Datensatz allerdings grösser ist, können wir nicht so leicht alle Werte einzeln korrigieren. In solch einem Fall können wir z.B. eine *obere Schranke* definieren, welche von keinem der Werte in einer Spalte überschritten werden sollte.

```python
for x in workout.index: # iteriert in einer Schleife über alle Indizes des DataFrame's workout
  if workout.loc[x, 'dauer'] > 120: # mit loc[x, y] kann auf einen bestimmten Wert zugegriffen werden
    workout.loc[x, 'dauer'] = 120 # ersetzt alle Einträge in der Spalte 'dauer' die grösser als 120 sind mit 120.
```

Anstatt diese Werte zu ersetzen, können wir auch alle diese Zeilen verwerfen.

```python
for x in workout.index:
  if workout.loc[x, 'dauer'] > 120:
    workout.drop(x, inplace = True) # die Funktion drop() entfernt die angegebene Zeile aus dem DataFrame
```

In unserem Fall passiert beim Ausführen der beiden Zellen oben nichts, da in der Spalte *dauer* keine Werte über 120 mehr vorkommen.
